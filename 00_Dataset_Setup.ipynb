```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic dataset matching Sprint requirements
n_samples = 1000

data = {
    "CustomerID": [f"CUST-{1000 + i}" for i in range(n_samples)],
    "Age": np.random.choice(
        [25, 34, 45, 52, 61, np.nan, 150, -5], size=n_samples, p=[0.2, 0.2, 0.2, 0.15, 0.15, 0.05, 0.02, 0.03]
    ),
    "Gender": np.random.choice(["Male", "Female", "M", "F", np.nan], size=n_samples, p=[0.4, 0.4, 0.08, 0.08, 0.04]),
    "TenureYears": np.random.choice([0, 1, 2, 3, 5, 8, 10], size=n_samples),
    "MonthlyCharges": np.random.choice(
        ["$29.85", "$56.95", "99.99", "$105.50", "INVALID", np.nan], size=n_samples, p=[0.3, 0.3, 0.2, 0.1, 0.05, 0.05]
    ),
    "TotalCharges": np.random.uniform(100.0, 8000.0, size=n_samples),
    "ContractType": np.random.choice(
        ["Month-to-month", "month to month", "m2m", "One year", "Two year", "Unknown"], size=n_samples
    ),
    "PaymentMethod": np.random.choice(
        ["Electronic check", "Mailed check", "Bank transfer", "Credit card"], size=n_samples
    ),
    "Churn": np.random.choice(["No", "Yes"], size=n_samples, p=[0.7, 0.3]),
}

df = pd.DataFrame(data)

# Introduce 10 duplicate rows to test duplicate handling
duplicates = df.iloc[:10].copy()
df = pd.concat([df, duplicates], ignore_index=True)

# Save to CSV in the current working directory
df.to_csv("Customer_Data.csv", index=False)

print("SUCCESS: Customer_Data.csv has been created in your directory!")
print(f"Dataset Shape: {df.shape}")
```
